```python
# Looking at one stock for sanity check
reliance = df[df["Symbol"] == "RELIANCE"].sort_values("Date")
print(reliance[["Date", "Close", "Volume", "Turnover"]].head(10))
```

```python
# Quick sanity check: Turnover should roughly equal Volume * Close
reliance['Implied_Turnover'] = reliance['Volume'] * reliance['Close']
print(reliance[['Implied_Turnover', 'Turnover']].head(10))
```

```python
reliance['ratio'] = reliance['Turnover'] / reliance['Implied_Turnover']
print(reliance['ratio'].describe())
```

```python
tcs = df[df["Symbol"] == "TCS"].sort_values("Date")
tcs['Implied_Turnover'] = tcs['Volume'] * tcs['Close']
tcs['ratio'] = tcs['Turnover'] / tcs['Implied_Turnover']
print(tcs['ratio'].describe())
```

#### Data Quality Finding: Turnover column

- Turnover values are ~100,000x too large across all stocks checked (Reliance, TCS)
- Verified by comparing Turnover to Volume × Close — ratio consistently ≈ 100,000
- Likely cause: source reports Turnover in ₹ Lakhs (1 lakh = 100,000), not converted correctly
- Fix: divide Turnover by 100,000 during ingestion

```python
# For each year, what fraction of rows have Trades missing?
df['Year'] = pd.to_datetime(df['Date']).dt.year
missing_by_year = df.groupby('Year')['Trades'].apply(lambda x: x.isna().mean())
print(missing_by_year)
```

```python
df.groupby('Year')['Trades'].count()
```

#### Data Quality Finding: Trades column

- Trades is 100% missing for 2000-2010, transitions in 2011 (~41% missing), 
  fully populated from 2012 onward
- Not random — a clean reporting cutoff, likely NSE started publishing this 
  field around 2011
- Decision: keep Trades as-is (NaN before 2012); exclude it from any analysis 
  spanning the full date range, only use it for 2012+ analysis
